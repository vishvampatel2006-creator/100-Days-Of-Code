```python
class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print(f"Current Balance: {account.get_balance()}")
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_from_origin(self):
        return math.sqrt(self.x**2 + self.y**2)

point = Point(3, 4)
print(f"Distance from (0,0): {point.distance_from_origin()}")
```

### [16] The Bank Loan Approval & Savings Hierarchy
A financial institution manages different types of user accounts:
- Create a base class `Account` with attributes `account_number` and `balance`.
- Derive a child class `SavingsAccount` that adds an attribute `interest_rate` and a method `add_interest()` that updates `balance`.
- Derive another child class `LoanAccount` from `Account` that adds `loan_amount` and a method `repay_loan(amount)` which reduces `loan_amount` and deducts from `balance`.
- Instantiate both account types and process a transaction workflow.

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

class SavingsAccount(Account):
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        self.balance += self.balance * (self.interest_rate / 100)

class LoanAccount(Account):
    def __init__(self, account_number, balance, loan_amount):
        super().__init__(account_number, balance)
        self.loan_amount = loan_amount

    def repay_loan(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            self.loan_amount -= amount

savings = SavingsAccount("SA001", 5000, 4.0)
savings.add_interest()
print(f"Savings Balance after interest: {savings.balance}")

loan = LoanAccount("LA001", 10000, 2000)
loan.repay_loan(500)
print(f"Loan Remaining: {loan.loan_amount}, Account Balance: {loan.balance}")
```

### [17] The E-Commerce Product Hierarchy Engine
An online store tracks physical and digital inventory:
- Define a base class `Product` with `product_id`, `title`, and `base_price`. Add a method `get_final_price()` that returns `base_price`.
- Create a derived class `PhysicalProduct(Product)` that adds `shipping_cost` and overrides `get_final_price()` to return `base_price + shipping_cost`.
- Create a derived class `DigitalProduct(Product)` that adds `discount_percent` and overrides `get_final_price()` to return price after discount without shipping fees.

```python
class Product:
    def __init__(self, product_id, title, base_price):
        self.product_id = product_id
        self.title = title
        self.base_price = base_price

    def get_final_price(self):
        return self.base_price

class PhysicalProduct(Product):
    def __init__(self, product_id, title, base_price, shipping_cost):
        super().__init__(product_id, title, base_price)
        self.shipping_cost = shipping_cost

    def get_final_price(self):
        return self.base_price + self.shipping_cost

class DigitalProduct(Product):
    def __init__(self, product_id, title, base_price, discount_percent):
        super().__init__(product_id, title, base_price)
        self.discount_percent = discount_percent

    def get_final_price(self):
        return self.base_price - (self.base_price * (self.discount_percent / 100))

p_prod = PhysicalProduct(1, "Keyboard", 50, 10)
d_prod = DigitalProduct(2, "E-Book", 30, 20)
print(f"Physical Price: {p_prod.get_final_price()}, Digital Price: {d_prod.get_final_price()}")
```

### [18] The University Role Management System
A university portal manages administrative privileges across campus roles:
- Create a base class `User` with private attributes `__username` and `__email`, providing getter methods.
- Inherit a class `Faculty(User)` that adds `subject_specializations` (a list) and a method `assign_course(course_name)`.
- Inherit a class `Student(User)` that adds `enrolled_courses` (a list) and a method `enroll(course_name)`.
- Demonstrate creating instances for both roles and securely updating their course records.

```python
class User:
    def __init__(self, username, email):
        self.__username = username
        self.__email = email

    def get_username(self):
        return self.__username

    def get_email(self):
        return self.__email

class Faculty(User):
    def __init__(self, username, email):
        super().__init__(username, email)
        self.subject_specializations = []

    def assign_course(self, course_name):
        self.subject_specializations.append(course_name)

class Student(User):
    def __init__(self, username, email):
        super().__init__(username, email)
        self.enrolled_courses = []

    def enroll(self, course_name):
        self.enrolled_courses.append(course_name)

prof = Faculty("prof_smith", "smith@univ.edu")
prof.assign_course("Data Structures")

student = Student("john_d", "john@univ.edu")
student.enroll("Data Structures")

print(f"Faculty {prof.get_username()} teaches {prof.subject_specializations}")
print(f"Student {student.get_username()} is enrolled in {student.enrolled_courses}")
```

### [19] The Smart Home IoT Device Controller
A smart home gateway controls connected devices:
- Create a parent class `SmartDevice` with attributes `device_id` and `status` (`"ON"`/`"OFF"`), with methods `turn_on()` and `turn_off()`.
- Create a child class `SmartThermostat(SmartDevice)` that adds `temperature` and a method `set_temperature(temp)` (allowing temperature changes only if `status` is `"ON"`).
- Create a child class `SmartLight(SmartDevice)` that adds `brightness_level` (0-100%).

```python
class SmartDevice:
    def __init__(self, device_id):
        self.device_id = device_id
        self.status = "OFF"

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

class SmartThermostat(SmartDevice):
    def __init__(self, device_id, temperature=72):
        super().__init__(device_id)
        self.temperature = temperature

    def set_temperature(self, temp):
        if self.status == "ON":
            self.temperature = temp

class SmartLight(SmartDevice):
    def __init__(self, device_id, brightness_level=0):
        super().__init__(device_id)
        self.brightness_level = brightness_level

thermostat = SmartThermostat("T01")
thermostat.turn_on()
thermostat.set_temperature(75)
print(f"Thermostat {thermostat.device_id} is {thermostat.status} at {thermostat.temperature} degrees.")
```

### [20] The Corporate Employee Payroll Pipeline
A company calculates monthly payouts using inheritance:
- Create a base class `Employee` with attributes `emp_id`, `name`, and a method `calculate_pay()` returning `0.0`.
- Derive `SalariedEmployee(Employee)` with attribute `monthly_salary`, overriding `calculate_pay()` to return the fixed salary.
- Derive `HourlyEmployee(Employee)` with attributes `hourly_rate` and `hours_worked`, overriding `calculate_pay()` to return `hourly_rate * hours_worked`.
- Store a list of mixed employee objects and iterate through them to compute total corporate payroll using polymorphism.

```python
class Employee:
    def __init__(self, emp_id, name):
        self.emp_id = emp_id
        self.name = name

    def calculate_pay(self):
        return 0.0

class SalariedEmployee(Employee):
    def __init__(self, emp_id, name, monthly_salary):
        super().__init__(emp_id, name)
        self.monthly_salary = monthly_salary

    def calculate_pay(self):
        return self.monthly_salary

class HourlyEmployee(Employee):
    def __init__(self, emp_id, name, hourly_rate, hours_worked):
        super().__init__(emp_id, name)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_pay(self):
        return self.hourly_rate * self.hours_worked

employees = [
    SalariedEmployee(101, "Alice", 6000),
    HourlyEmployee(102, "Bob", 25, 120)
]

total_payroll = sum(emp.calculate_pay() for emp in employees)
print(f"Total Corporate Payroll: ${total_payroll}")
```
