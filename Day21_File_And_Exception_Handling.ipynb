```python
filename = "students.txt"

# 1. Writing to a text file
try:
    file = open(filename, "w")
    file.write("RollNo, Name, Marks\n")
    file.write("101, Vishva, 88\n")
    file.write("102, Maanay, 92\n")
except IOError as e:
    print("File write karne mein error aaya:", e)
else:
    print("Data successfully write ho gaya!")
finally:
    file.close()
    print("Write operation ke baad file close ho gayi.\n")

# 2. Reading from a file (with FileNotFoundError handling)
def read_file(fname):
    try:
        with open(fname, "r") as f:
            content = f.read()
    except FileNotFoundError:
        print(f"Error: '{fname}' naam ki koi file nahi mili!")
    else:
        print(f"--- '{fname}' ka content ---")
        print(content)
    finally:
        print("Read attempt complete.\n")

read_file("students.txt")   # Sahi file
read_file("missing.txt")    # Error test karne ke liye
```

```python
import csv

csv_file = "marks.csv"
data = [
    ["Subject", "Code", "Marks"],
    ["Python", "BTCS305", 38],
    ["DSA", "BTCS202", 35],
    ["Digital Electronics", "BTCS205", 34]
]

# Writing CSV
try:
    with open(csv_file, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerows(data)
    print("CSV file ban gayi!")
except Exception as e:
    print("CSV write error:", e)

# Reading CSV & Calculating Average Marks
try:
    with open(csv_file, "r") as f:
        reader = csv.reader(f)
        header = next(reader)  # First row (header) skip karne ke liye
        total = 0
        count = 0
        for row in reader:
            print(row)
            total += int(row[2])  # ValueError aa sakta hai agar number na ho
            count += 1
        print(f"\nAverage Marks: {total / count:.2f}")
except FileNotFoundError:
    print("CSV file nahi mili!")
except ValueError:
    print("Marks column mein invalid number hai!")
except ZeroDivisionError:
    print("File mein koi data rows nahi hain!")
finally:
    print("CSV processing khatam.")
```
